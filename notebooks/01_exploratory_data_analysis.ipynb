## Project: AI-Powered Energy Consumption Forecasting

### Purpose
This notebook serves as the initial exploratory environment to inspect the statistical distributions, temporal patterns, seasonality, and sub-metering characteristics of the Individual Household Electric Power Consumption dataset.

> **Prerequisite**: Before running this notebook, download the raw dataset from the [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/235/individual+household+electric+power+consumption) and place `household_power_consumption.txt` inside `../data/raw/`.

### Planned Analysis Steps

1. **Data Ingestion & Integrity Check**:
   - Parse semicolon-separated records and construct a continuous timestamp index.
   - Quantify missing value rates (marked as `?` in raw observations).

2. **Summary Statistics & Distribution Analysis**:
   - Examine statistical summaries (mean, standard deviation, quantiles) for `Global_active_power`, `Voltage`, and `Global_reactive_power`.
   - Detect anomalies and sensor dropouts.

3. **Time-Series Decomposition & Periodicity**:
   - Resample series to hourly and daily intervals.
   - Analyze daily load profiles (morning and evening peaks) and weekday vs. weekend variances.
   - Evaluate seasonal trends across winter and summer months.

4. **Sub-Metering Correlation Analysis**:
   - Compare sub-metering breakdown: Kitchen (`Sub_metering_1`), Laundry (`Sub_metering_2`), and Climate control (`Sub_metering_3`).
   - Calculate residual active energy not captured by sub-meters.

5. **Insights for Feature Engineering**:
   - Identify optimal lag windows (e.g., 24h, 168h) and required rolling window aggregations.
